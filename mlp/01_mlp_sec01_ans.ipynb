### 例題2. ニューラルネットワーククラスの定義とインスタンスの宣言

#### 例題2のコード

````
# 例題2. ニューラルネットワーククラスの定義とインスタンスの宣言

# 1. ニューラルネットワークの定義
class BeafClassifier(nn.Module):
    def __init__(self):
        super(BeafClassifier, self).__init__()
        self.fc1 = nn.Linear(4, 5)

    def forward(self, x):
        x = self.fc1(x)
        return x

# 2. インスタンスの宣言
beaf_classifier = BeafClassifier()
````

### 例題5. 誤差逆伝播とパラメータの更新

教師データとの誤差を計算し，パラメータを更新する．

---



#### 例題5のコード

````
# 例題5. 誤差逆伝播とパラメータの更新

epochs = 20
for epoch in range(epochs):
    print("学習{}回目".format(epoch+1))
  
    # 1. パラメータの微分値を初期化
    optimizer_beaf_classifier.zero_grad()
  
    # 2. 誤差の計算
    loss = criterion(outputs, labels)

    # 3. 誤差逆伝播
    loss.backward()

    # 4. パラメータの更新
    optimizer_beaf_classifier.step()

    # 更新後のニューラルネットワークから出力
    outputs = beaf_classifier(inputs)

    print("AIからの実測値　：", outputs.data)
    print("予測ランクはA{}\n".format(5-outputs.argmax()))
````

#### 1. パラメータの微分値を初期化  


- パラメータの微分値の初期化はoptimパッケージ内のクラスインタンス（最適化器）が持つ


# まとめ

スライドで学習したNNの学習の原理とPyTorchの記述方法を比較してみます．  
NNの学習の原理として，1．順伝播，2．誤差計算，3．誤差逆伝播，4．最適化の順番で処理を行っていくと解説しました．  
PyTorchの処理をこれに対応させると，1．NNに入力を与える，2．NNの出力と教師データを誤差関数に与える，3．誤差逆伝播，4．パラメータの更新となります．  
このNNの学習の原理で行う処理の他にNNクラスの定義やデータセットの読み込みなどといった前処理がありました．  
この流れは基本形になりますので，しっかりと覚えておきましょう．
